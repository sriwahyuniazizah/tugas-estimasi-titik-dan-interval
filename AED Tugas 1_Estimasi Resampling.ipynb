```python
import numpy as np

np.random.seed(42)
n_sampel = 100
data = np.random.normal(loc=50, scale=8, size=n_sampel)
print(f"Jumlah sampel: {len(data)}")
print(f"Data pertama (5 data teratas): {data[:5]}\n")
```

```python
from scipy import stats

print("--- BAGIAN A: Estimasi Titik & Interval (Distribusi Z) ---")
point_estimate = np.mean(data)
print(f"1. Nilai Estimasi Titik (Mean Sampel): {point_estimate:.4f}")
alpha = 0.05
confidence_level = 1 - alpha
s_sampel = np.std(data, ddof=1)
z_score = stats.norm.ppf(1 - alpha / 2)
margin_of_error = z_score * (s_sampel / np.sqrt(n_sampel))
lower_limit = point_estimate - margin_of_error
upper_limit = point_estimate + margin_of_error
print(f"   Standar Deviasi Sampel (s): {s_sampel:.4f}")
print(f"   Nilai Z (untuk alpha 5% / 95% confidence): {z_score:.4f}")
print(f"   Margin of Error: {margin_of_error:.4f}")
print(f"2. Estimasi Interval ({int(confidence_level*100)}%): "
      f"({lower_limit:.4f} <= µ <= {upper_limit:.4f})\n")
```

```python
print("--- BAGIAN B: Resampling (Jackknife & Bootstrapping) ---")
thetas_jackknife = np.empty(n_sampel)
for i in range(n_sampel):
    data_deleted = np.delete(data, i)
    thetas_jackknife[i] = np.mean(data_deleted)

mean_jackknife = np.mean(thetas_jackknife)
print(f"1. Jackknife:")
print(f"   - Rata-rata dari hasil resampling Jackknife: {mean_jackknife:.4f}")

n_iterations = 1000
thetas_bootstrap = np.empty(n_iterations)

for b in range(n_iterations):
    boot_sample = np.random.choice(data, size=n_sampel, replace=True)
    thetas_bootstrap[b] = np.mean(boot_sample)

mean_bootstrap = np.mean(thetas_bootstrap)
print(f"2. Bootstrapping (1000 iterasi):")
print(f"   - Rata-rata dari hasil resampling Bootstrap: {mean_bootstrap:.4f}")
```
